```python
import numpy as np
import pandas as pd
from src.metrics import evaluate_regression
from src.models.baseline import MeanBaseline
import time
```

```python
data = pd.read_csv("//dataset/staging/data_sau_clean.csv")

print("Shape:", data.shape)
```

```python
from src.feature_engineering.build_feature import build_features

features = build_features(
    data
)

X_train = features["X_train"]
X_test  = features["X_test"]
y_train = features["y_train"]
y_test  = features["y_test"]
```

```python
from sklearn.model_selection import train_test_split

# Chia train -> train + val
X_train, X_val, y_train, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.2,      # 20% cho validation
    random_state=42,
    shuffle=True
)

print("Train:", X_train.shape, y_train.shape)
print("Val  :", X_val.shape, y_val.shape)
print("Test :", X_test.shape, y_test.shape)
```

```python
results = {}

baseline = MeanBaseline()

start = time.time()
baseline.fit(y_train)
train_time = time.time() - start

y_pred = baseline.predict(y_test)

results["Baseline"] = {
    **evaluate_regression(y_test, y_pred),
    "train_time": train_time
}
```

```python
print(results)
```
